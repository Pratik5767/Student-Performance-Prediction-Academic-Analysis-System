### Clean the data

```python
clean = df.copy()
```

```python
# removing duplicates
row_count = len(clean)

clean = clean.drop_duplicates()
print('Duplicates removed:', row_count - len(clean))
```

```python
# handling missing values
for col in ["Attendance", "Assignment_Score"]:
    median = clean[col].median()
    clean[col] = clean[col].fillna(median)
    print(f"Filled missing '{col}' with median = {median}")
```

```python
# checking everthing is correct
print("Missing values after cleaning:\n", clean.isnull().sum())
print("Duplicates after cleaning:", clean.duplicated().sum())
print("Final number of students:", len(clean))
```

### Analyze student performance

- Question 1 - Do students with higher attendance generally score better?

```python
print(clean.groupby('Final_Result')["Attendance"].mean().round(1))
```

- Question 2 - Do students who study more hours generally perform better?

```python
print(clean.groupby('Final_Result')["Study_Hours"].mean().round(1))
```

- Question 3 - Does previous performance affect the final result?

```python
print(clean.groupby('Final_Result')['Previous_Marks'].mean().round(1))
```

- Question 4 - Do students with previous failures have a higher chance of failing?

```python
pass_rate_table = pd.crosstab(clean['Previous_Failures'], clean['Final_Result'], normalize='index') * 100
print(pass_rate_table.round(1))
```

- Question 5 - Which factor appears most related to student performance?

```python
features = ["Study_Hours", "Attendance", "Previous_Marks", "Assignment_Score", "Internal_Marks", "Previous_Failures"]

temp_target = clean["Final_Result"].map({"Pass": 1, "Fail": 0})
correlations = clean[features].corrwith(temp_target).sort_values(ascending=False)
print(correlations.round(2))
```

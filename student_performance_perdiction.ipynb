#### Clean the data

```python
clean = df.copy()
```

```python
# removing duplicates
row_count = len(clean)

clean = clean.drop_duplicates()
print('Duplicates removed:', row_count - len(clean))
```

### Prepare Data for ML

```python
clean['Final_Result'] = clean['Final_Result'].map({"Pass": 1, "Fail": 0})

print(clean['Final_Result'].value_counts())

df.head()
```

### Select input features

```python
X = clean[
    [
        "Study_Hours",
        "Attendance",
        "Previous_Marks",
        "Assignment_Score",
        "Internal_Marks",
        "Previous_Failures"
    ]
]

y = clean["Final_Result"]

print("X shape", X.shape)
print("y shape", y.shape)
```

### Train/Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training students: ", X_train.shape[0])
print("Testing students: ", X_test.shape[0])
```

### Build a Simple ML Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

print("Model trained successfully")
```

### Make Predictions

```python
predictions = model.predict(X_test)

print(predictions)
```

```python
compare = pd.DataFrame({
    "Actual": y_test.values[:10],
    "Predicted": predictions[:10]
})

compare['Correct'] = compare["Actual"] == compare["Predicted"]

print(compare)
```

### Evaluate the Model

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test, predictions)

print("Accuracy: ", round(accuracy * 100, 2), "%")
```

### Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, predictions)

print(cm)
```

### Optional Second Model

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model2 = DecisionTreeClassifier(random_state=42)

model2.fit(X_train, y_train)

predictions2 = model2.predict(X_test)
accuracy2 = accuracy_score(y_test, predictions2)
```

```python
comparison_table = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy (%)": [round(accuracy * 100, 2), round(accuracy2 * 100, 2)]
})

comparison_table
```

### Saving trained model

```python
joblib.dump(model, "student_pass_fail_model.pkl")

print("Model saved to student_pass_fail_model.pkl")
```

```python
# load it back and predict once
loaded_model = joblib.load("student_pass_fail_model.pkl")

print("Reloaded model accuracy on test data:", round(accuracy_score(y_test, loaded_model.predict(X_test)) * 100, 2), "%")
```

### Simple Recommendation System

```python
features = [
    "Study_Hours", 
    "Attendance", 
    "Previous_Marks",             
    "Assignment_Score", 
    "Internal_Marks", 
    "Previous_Failures"
]
```

```python
def student_performance_recommendation(study_hours, attendance, previous_marks, assignment_score, internal_marks, previous_failures):
    new_student = pd.DataFrame(
        [[study_hours, attendance, previous_marks, assignment_score, internal_marks, previous_failures]],
        columns=features
    )

    result = model.predict(new_student)[0]
    probability = model.predict_proba(new_student)[0][1] * 100

    # 3) print a neat report
    print("=" * 40)
    print("     STUDENT PERFORMANCE PREDICTION")
    print("=" * 40)
    print()
    print(f"Study Hours       : {study_hours}")
    print(f"Attendance        : {attendance}")
    print(f"Previous Marks    : {previous_marks}")
    print(f"Assignment Score  : {assignment_score}")
    print(f"Internal Marks    : {internal_marks}")
    print(f"Previous Failures : {previous_failures}")
    print()
    if result == 1:
        print("Predicted Result  : PASS")
        print(f"Chance of passing : {probability:.1f}%")
        print()
        print("Recommendation:")
        print("Student performance is satisfactory.")
        print("Continue current study pattern.")
    else:
        print("Predicted Result  : FAIL")
        print(f"Chance of passing : {probability:.1f}%")
        print()
        print("Recommendation:")
        print("Student requires additional academic support.")
        print("Improve attendance and study hours.")
    print("=" * 40)
```

```python
student_performance_recommendation(study_hours=3, attendance=68, previous_marks=52, assignment_score=60, internal_marks=55, previous_failures=1)
```
